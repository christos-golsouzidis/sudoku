```python
weights = np.array([])
weights = np.array(range(0,9))
weights = 10**weights
weights = np.insert(weights, 0, 0)
weights
```

```python
def winones(obj, position):
    x = int(position / 3)
    for x in range(3*x,3*x+3):
        y = int(position % 3)
        for y in range(3*y,3*y+3):
            obj[x,y] = 1
```

```python
window = np.zeros([9,9], dtype=np.int8)
winones(window, 0)
window
```

```python
def testwin(obj, position):
    window = np.zeros([9,9], dtype=np.int8)
    winones(window, position)
    window = window * -1
    z = obj & window
    return z
```

```python
def vmatch(obj, row):
    m=0
    while(m < row):
        for x in range(0,9):
            if(obj[row][x] == obj[m][x]):
                return False
        m += 1
    return True
```
